# Arabic Legal RAG: Hybrid Retrieval & Evaluation Walkthrough
This notebook provides a complete reproducible walkthrough of:
1. Loading the statutory Egyptian Labor Law corpus (`data/egypt_labor_law_corpus.jsonl`)
2. Arabic text normalization (Tashkeel removal, Tatweel stripping, Hamza unification)
3. Sparse lexical indexing using BM25Okapi
4. Dense semantic retrieval with `multilingual-e5-base`
5. Reciprocal Rank Fusion (RRF) and Cross-Encoder re-ranking evaluation

```python
import json
from pathlib import Path
from src.preprocessor.arabic_cleaner import ArabicTextNormalizer
from src.retriever.bm25_retriever import ArabicBM25Retriever
from src.retriever.hybrid import ReciprocalRankFusion

# Load real corpus
corpus_path = Path('../data/egypt_labor_law_corpus.jsonl')
with open(corpus_path, 'r', encoding='utf-8') as f:
    documents = [json.loads(line) for line in f]
print(f'Successfully loaded {len(documents)} statutory legal articles.')
```

```python
# Initialize normalizer and sparse index
normalizer = ArabicTextNormalizer()
bm25 = ArabicBM25Retriever()
bm25.index(documents)

# Query test
query = 'ما هي شروط انتهاء عقد العمل محدد المدة؟'
results = bm25.search(query, top_k=3)
for r in results:
    print(f"Rank #{r['rank']} | Score: {r['score']:.4f} | Chunk: {r['chunk_id']}")
    print(f"  {r['text'][:120]}...")
```
